## 답안 작성을 위해 미리 작성해둘 내용

```python
# 답안 Template - 공통적으로 미리 작성해놓기
common_result = {}
common_result['\n[문제 번호]'] = '1.2\n'
common_result['key1'] = 'value1'
common_result['key2'] = 'value2'
common_result['key3'] = 'value3'
for k, v in common_result.items(): print("{} : {}".format(k, v))

# 답안 Template - 통계 파트 미리 작성해놓기
fc = lambda x : x + 5

test_result = {}
test_result['\n[문제 번호]'] = '4.4\n'
test_result['귀무가설'] = '모집단의 평균은 0이다.'
test_result['대립가설'] = '모집단의 평균은 0이 아니다.'
test_result['검정통계량'] = fc(4)
test_result['p-value'] = '0.04'
test_result['결과 해석'] = 'p-value가 0.05보다 작기에, 유의수준 하에서 귀무가설을 기각하는 것이 통계적으로 유의하다.'
for k, v in test_result.items(): print("{} : {}".format(k, v))

```
