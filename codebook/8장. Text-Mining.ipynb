```python
import numpy as np
from src.TextMining.text import TextProfile, TopicProfile, run_vectorization, run_text_classification, run_topic_model
texts = [('good excellent happy' if i%2 else 'bad awful sad')+' unique'+str(i) for i in range(40)]
labels = np.arange(40)%2
vectors = run_vectorization(texts[:30], test_documents=texts[30:])
print(vectors.train_matrix.shape, vectors.test_matrix.shape)
print(vectors.feature_names)
```

```python
result = run_text_classification(texts, labels, TextProfile(param_grid={'model__C':[0.1,1]}, cv=3))
print(result.test_metrics, result.best_params)
topics = run_topic_model(texts, TopicProfile(max_iter=5))
print(topics.top_words)
print(topics.document_topics[:3])
```
