```python
import numpy as np
import pandas as pd
from src.TimeSeries.forecasting import ForecastProfile, SeriesDiagnosticsProfile, run_forecast, run_series_diagnostics
series = pd.Series(20+np.cumsum(np.random.RandomState(4).normal(size=80)), index=pd.date_range('2020-01-01', periods=80, freq='D'))
# 정상성 진단도 보류 구간을 제외한 학습 데이터로 수행합니다.
diagnostics = run_series_diagnostics(series.iloc[:-8], SeriesDiagnosticsProfile(difference_order=1))
print(diagnostics['adf_p_value'], diagnostics['reject_unit_root'])
```

```python
result = run_forecast(series, ForecastProfile(holdout=8, order_candidates=[(0,1,0), (1,1,0)]))
print(result.best_order, result.metrics)
print(result.predictions)
print(result.intervals)
print(result.candidates)
```
