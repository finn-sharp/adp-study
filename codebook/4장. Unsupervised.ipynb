```python
from sklearn.datasets import load_iris
from src.MachineLearning.Clustering.clustering import ClusteringProfile, run_clustering, make_linkage_matrix
X, y = load_iris(return_X_y=True)
for method in ['kmeans', 'hierarchical', 'dbscan', 'gaussian_mixture']:
    result = run_clustering(X, ClusteringProfile(method=method, n_init=2), labels_true=y)
    print(method, result.n_clusters, result.n_noise, result.metrics)
```

```python
from src.MachineLearning.Decomposition.pca import DecompositionProfile, run_decomposition
pca = run_decomposition(X)
print(pca.transformed.shape, pca.cumulative_variance_ratio)
factor = run_decomposition(X, DecompositionProfile(method='factor_analysis'))
print(factor.transformed.shape)
```
