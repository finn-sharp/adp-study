```python
from src.stats.analysis import StatisticalProfile, run_descriptive, run_statistical_test
from src.stats.tests.parametric.Average.Univariate import TestProfile, run_one_sample_test
from src.stats.tests.parametric.Average.Bivariate import TwoSampleTestProfile, run_two_sample_test
from src.stats.tests.parametric.Average.Multivariate.OneWay import OneWayANOVAProfile, run_oneway_anova
a, b, c = [1,2,4,5,7,8], [2,3,3,6,7,10], [10,11,12,13,14,16]
print(run_descriptive(a))
print(run_one_sample_test(a, TestProfile(mu_0=3)))
print(run_two_sample_test(a, b, TwoSampleTestProfile(equal_var=False)))
print(run_oneway_anova([a,b,c], OneWayANOVAProfile(equal_var=False)).anova_table)
```

```python
print(run_statistical_test(a, StatisticalProfile(test='paired_t'), other=b))
print(run_statistical_test([a,b,c], StatisticalProfile(test='kruskal')))
print(run_statistical_test([[10,20],[30,40]], StatisticalProfile(test='chi2_independence')))
print(run_statistical_test(a, StatisticalProfile(test='variance_chi2', null_value=4)))
```
